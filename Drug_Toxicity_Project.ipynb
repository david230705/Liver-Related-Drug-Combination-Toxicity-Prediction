```python
# ============================================================
# CELL 1 — PROJECT SETUP
# Run this first after starting/restarting Colab
# ============================================================

from google.colab import drive
drive.mount("/content/drive")

import os
import sys
import subprocess
import time
import requests

PROJECT_PATH = "/content/drive/MyDrive/Drug-Toxicity-Project"
SRC_PATH = PROJECT_PATH + "/src"
APP_PATH = PROJECT_PATH + "/app"

# Add project source folder
if SRC_PATH not in sys.path:
    sys.path.insert(0, SRC_PATH)

print("==========================================")
print("Drug Toxicity Project")
print("==========================================")
print("Project:", PROJECT_PATH)

# Check important files
required_files = [
    "data/processed/liver_toxicity_pairs_with_smiles.csv",
    "data/processed/validated_drug_smiles.csv",
    "data/processed/kg_embeddings_rxnorm.pt",
    "data/processed/molecular_graph_records.pkl",
    "outputs/checkpoints/full_model_cold_best.pt",
    "app/app.py",
    "app/index.html",
    "app/drugs.html"
]

print("\nChecking project files...\n")

all_ok = True

for file in required_files:

    path = os.path.join(PROJECT_PATH, file)

    if os.path.exists(path):
        print("✅", file)
    else:
        print("❌ MISSING:", file)
        all_ok = False

print("\n==========================================")

if all_ok:
    print("✅ Project files are ready.")
else:
    print("⚠️ Some required files are missing.")

print("==========================================")
```

```python
# ============================================================
# CELL 2 — START BACKEND
# ============================================================

import subprocess
import time
import requests

PROJECT_PATH = "/content/drive/MyDrive/Drug-Toxicity-Project"

# Stop any old FastAPI server
subprocess.run(
    "pkill -f 'uvicorn app.app:app'",
    shell=True,
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

time.sleep(2)

# Start FastAPI backend
server = subprocess.Popen(
    [
        "python", "-m", "uvicorn",
        "app.app:app",
        "--host", "0.0.0.0",
        "--port", "8000"
    ],
    cwd=PROJECT_PATH,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

print("Starting backend...")

# Wait for backend
backend_ready = False

for _ in range(30):
    time.sleep(1)

    if server.poll() is not None:
        break

    try:
        response = requests.get(
            "http://127.0.0.1:8000/",
            timeout=2
        )

        if response.status_code == 200:
            backend_ready = True
            break

    except requests.exceptions.RequestException:
        pass

print()

if backend_ready:
    print("==========================================")
    print("✅ BACKEND RUNNING")
    print("==========================================")
    print("FastAPI: http://127.0.0.1:8000")

else:
    print("❌ Backend failed to start.")
    print("\n========== BACKEND LOGS ==========")

    if server.poll() is not None:
        print(server.stdout.read())
    else:
        print("Server is still running. Check the endpoint.")
```

```python
# ============================================================
# CELL 3 — OPEN PROJECT
# ============================================================

from google.colab.output import eval_js

print("==========================================")
print("🌐 DRUG TOXICITY PREDICTION SYSTEM")
print("==========================================")

# Create Colab public proxy URL
public_url = eval_js(
    "google.colab.kernel.proxyPort(8000)"
)

print()
print("OPEN THIS LINK:")
print()
print(public_url)
print()
print("==========================================")
print("✅ Frontend + Backend are ready")
print("==========================================")
```

```python
# ============================================================
# CELL 3 — STOP FASTAPI BACKEND
# ============================================================

if server and server.poll() is None:
    server.terminate()
    server.wait(timeout=10)

    print("==========================================")
    print("✅ FASTAPI BACKEND STOPPED")
    print("==========================================")
else:
    print("Backend is already stopped.")
```
